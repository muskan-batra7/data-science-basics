```python
import pandas as pd
```

```python
record= pd.read_csv("Titanic-Dataset.csv")

#gives number of missing (Null) values in each column
print(record["Age"].apply(type).value_counts())  
record.isnull().sum()
```

```python
dropped_rows=record.dropna() ## drop ANY row that has at least one missing value 
#same as record.dropna(axis="index",how="any")
#how=any-delete even if any value missing
#how=all-delete only even all values missing
#axis=index-drop rows
#axis=columns-drop columns

print("Before dropna():",len(record))
print("After dropna():",len(dropped_rows))

dropped_columns=record.dropna(axis="columns",how="any") #drop columns 
print(len(dropped_columns))


# drop rows only if a SPECIFIC column is missing
dropped_specified=record.dropna(subset=["Age","Name"])
print(len(dropped_specified))  #either age column is missing or name column is missing

# drop a COLUMN entirely if it has too many missing values
dropped_cabin=record.drop(columns="Cabin")  #drop cabin column
print(dropped_cabin.columns)

#to permanently apply changes
#record.dropna(inplace=True)
```

```python
#flling missing values

# fill missing Age with the median age
med=record["Age"].median()
record["Age"]=record["Age"].fillna(med)  #fill all the null values with the median value
print(record["Age"].isnull().sum())   #output is 0 as no value is null now
```

```python

#removing duplicates
print("Duplicated rows:",record.duplicated().sum())
record_no_dupes=record.drop_duplicates()  #remove all duplicated rows
print(len(record_no_dupes))

# check duplicates based on a SPECIFIC column only
print(record["Name"].duplicated().sum())
```

```python
#fixinf data types
print(record.dtypes)
#converting data type pf survived column from int to category
record["Survived"]=record["Survived"].astype("category")
print(record.dtypes)

```
