```python
import numpy as np 
import pandas as pd 
record=pd.read_csv("Titanic-Dataset.csv")
```

```python
# simulate a "booking date" column for practice
np.random.seed(7)
random_days=np.random.randint(0,60,size=len(record))
record["BookingDate"]=pd.to_datetime("1912-01-01")+pd.to_timedelta(random_days,unit="D")
print(record[["Name","BookingDate"]].head())
print(record["BookingDate"].dtype)  
```

```python
#Extracting parts of a date
record["BookingYear"]=record["BookingDate"].dt.year
record["BookingMonth"]=record["BookingDate"].dt.month
record["BookingDay"]=record["BookingDate"].dt.day
record["BookingWeekDay"]=record["BookingDate"].dt.day_name()
print(record[["BookingDate","BookingYear","BookingMonth","BookingDay","BookingWeekDay"]])
#oldest and newest date
print(record["BookingDate"].min())
print(record["BookingDate"].max())
```

```python
# passengers who booked in January
filt_month=record["BookingDate"].dt.month==1
print(record.loc[filt_month])

filt_day=record["BookingDate"].dt.day>15
print(record.loc[filt_day])


```

```python
# passengers who booked in the first two weeks
record["EarlyBooking"]=record["BookingDate"]<"1912-01-15"
print(record[["BookingDate","EarlyBooking"]])
```

```python
#resample
#filter data at different intervals
resample=record.set_index("BookingDate")["Survived"].resample("D").mean()
print(resample)
```

```python
# find passengers whose name contains "William"
williams=record[record["Name"].str.contains("William",case=False,na=False)]
print(williams[["Name"]].head())
```

```python
#split Name into Surname and FirstNames using the comma
record[["FirstName","LastName"]]=record["Name"].str.split(",",n=1,expand=True)
print(record[["Name","FirstName","LastName"]].head())
# strip extra whitespace that often comes from splitting
record["FirstName"]=record["FirstName"].str.strip()
record["LastName"]=record["LastName"].str.strip()
print(record[["FirstName","LastName"]])
```

```python
#some other string methods
print(record["Name"].str.upper().head())
print(record["Name"].str.lower().head())
# length of each name string
print(record["Name"].str.len().head())
#sum of names starting with B
print(record["Name"].str.startswith("B").sum())
```
