```python
import pandas as pd
```

```python
record=pd.read_csv("Titanic-Dataset.csv")
```

```python
# quick re-clean, same as Day 21
med=record["Age"].median()
record["Age"]=record["Age"].fillna(med)
# record=record.drop(columns=["Cabin"])
most_common_Port=record["Embarked"].mode()[0]
record["Embarked"]=record["Embarked"].fillna(most_common_Port)
```

```python
#groupby a single column
grouped=record.groupby("Pclass")
print(grouped.groups.keys())   #dict_keys([1, 2, 3])

# average survival rate per class
print(record.groupby("Pclass")["Survived"].mean())  #first groups by pclass then gives mean of survived for each pclass(1,2,3)

# average age per class
print(record.groupby("Pclass")["Age"].mean())

# count of passengers per class
print(record.groupby("Pclass")["Survived"].count())
```

```python
# group by class AND gender together
grouped=record.groupby(["Pclass","Sex"])
print(grouped.head())
print(grouped.get_group((1,"female")))
print(grouped["Survived"].mean())

# same thing, but as a proper DataFrame instead of a Series
survival_by_class_by_sex=record.groupby(["Pclass","Sex"])["Survived"].mean().reset_index()
print(survival_by_class_by_sex)
```

```python
#Multiple aggregations at once with .agg()
grp=record.groupby("Pclass")
summary=grp["Age"].agg(["mean","median","count"])
print(summary)

rec=record.groupby("Pclass").agg({
    "Survived":"mean",
    "Age":["mean","median"],
    "PassengerId":"count",
    "Fare":"mean"
})
print(rec)
```

```python
# sort grouped results to see which group had the best survival rate
survival_by_class_by_sex_sorted=survival_by_class_by_sex.sort_values("Survived",ascending=False)
print(survival_by_class_by_sex_sorted)

# sort the whole original DataFrame by a column
print(record.sort_values("Age",ascending=False).head()) # oldest passengers first

print(record.sort_values(["Pclass","Fare"],ascending=[True,False]).head())   # sort by multiple columns
```

```python
# does having family aboard affect survival?
record["FamilySize"]=record["SibSp"]+ record['Parch']
print(record.groupby("FamilySize")["Survived"].mean().sort_values(ascending=False))
```
